## Lectura del modelo

```python
modelo = joblib.load('/content/drive/MyDrive/modelo_colesterol.pkl')
```

## Uso:

```python
'''
horas_sueño
peso
horas_ejercicio
edad
genero (genero_Hombre, genero_Mujer)
'''
horas_sueño = int(input('Horas de sueño:'))
peso = int(input('Peso:'))
horas_ejercicio = int(input('Horas de ejercicio:'))
edad = int(input('Edad:'))
genero = input('Genero (H/M):')
genero_hombre = (genero.strip().upper() == 'H')
genero_mujer = (genero.strip().upper() == 'M')

nuevo_paciente = pd.DataFrame({
    'horas_sueño': [horas_sueño],
    'peso': [peso],
    'horas_ejercicio': [horas_ejercicio],
    'edad': [edad],
    'genero_Hombre': [genero_hombre],
    'genero_Mujer': [genero_mujer]
})

colesterol_predicho = modelo.predict(nuevo_paciente)
print(f'El colesterol predicho es: {colesterol_predicho[0]}')
```
