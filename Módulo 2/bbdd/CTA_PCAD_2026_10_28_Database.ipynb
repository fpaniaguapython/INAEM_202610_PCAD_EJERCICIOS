## Imports

```python
import sqlite3

import pandas as pd
```

## Establecimiento de la conexión

```python
conexion = sqlite3.connect('database.db')
```

## Obtención de los dataframes

```python
df_ccaa = pd.read_sql_query('SELECT * FROM comunidades_autonomas', conexion)
display(df_ccaa)
```

```python
df_provincias = pd.read_sql_query('SELECT * FROM provincias', conexion)
display(df_provincias.head())
```

```python
df_empleados = pd.read_sql_query('SELECT * FROM employees', conexion)
display(df_empleados.head())
```

## Obtención del nombre, la profesión y el salario de los empleados

```python
df_empleados = pd.read_sql_query('SELECT name, job_category, salary FROM employees', conexion)
display(df_empleados.head())
```

## Cierre de la conexión

```python
conexion.close()
```
